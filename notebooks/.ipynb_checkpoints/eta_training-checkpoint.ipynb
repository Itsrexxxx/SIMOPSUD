# SIMKO — Pelatihan Prediksi ETA & ETD (Dummy Realistis)

Dataset: `notebooks/data/` dari `generate_training_data.py`.

- **42 Lanud unik Kodau I–III** (GPS/ICAO nyata). Duplikat dibuang: Iskandar (I=II), Moeljono=Muljono, Dumatubun/DWT=Langgur.
- **30 tail TNI AU** + `CRUISE_SPEED_LOOKUP` (`f-16` 850, `sukhoi` 900, `t-50i` 750, `c-130` 540, `boeing` 780, `tucano` 480 km/h).
- **ETE deterministik, bukan random:** `minutes = haversine(origin,dest)/GS*60 * weather_severity + taxi`. GS = `cruise - 0.35*avg_wind`. Rute hanya dalam `MAX_RANGE` + `min_runway`.
- Katalog rute VFR: `route_ete_catalog.csv` (678 pasangan pesawat-siap × dest). Misi historis: 5000 baris `flight_missions_*.csv`.
- Model ETA = `RandomForestRegressor(distance, cruise_speed, avg_wind, severity)` — sama `eta_prediction_model.py`.
- Bottleneck = `RandomForestClassifier(avg_hours, trucks, forklifts, deployed)` — sama `predictive_model.py`.

````python
import pandas as pd, numpy as np, pathlib
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestRegressor, RandomForestClassifier
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score, classification_report, confusion_matrix

DATA = pathlib.Path("data")
df = pd.read_csv(DATA / "flight_missions_train.csv")
df["etd"] = pd.to_datetime(df["etd"])
print(f"train rows (Arrived): {len(df)}")
df[["tail_number","aircraft_type","origin_code","dest_code","distance_km","predicted_duration_minutes","actual_duration_minutes","delay_minutes","origin_weather","dest_weather","status"]].head(5)
````

````python
import matplotlib.pyplot as plt
import seaborn as sns

fig, ax = plt.subplots(1,2, figsize=(12,4))
sns.histplot(df["actual_duration_minutes"], bins=40, ax=ax[0]); ax[0].set_title("Actual duration (min)")
sns.histplot(df["delay_minutes"], bins=40, ax=ax[1]); ax[1].set_title("Delay vs ETA_pred (min)"); ax[1].axvline(0, color="red", ls="--")
plt.tight_layout(); plt.show()

print(df[["distance_km","cruise_speed_kmh","avg_wind_kmh","weather_severity","actual_duration_minutes","delay_minutes"]].describe().round(2).to_string())
print("\nTop routes:", (df["origin_code"]+"-"+df["dest_code"]).value_counts().head(8).to_dict())
print("By type:", df["type_key"].value_counts().to_dict())
````

## 1. Model ETA — prediksi durasi terbang (menit)

````python
# Fitur SAMA dengan eta_prediction_model.py: distance, cruise_speed, avg_wind, weather_severity
FEATURES = ["distance_km", "cruise_speed_kmh", "avg_wind_kmh", "weather_severity"]
TARGET = "actual_duration_minutes"

X = df[FEATURES].values
y = df[TARGET].values
Xtr, Xte, ytr, yte, dtr, dte = train_test_split(X, y, df, test_size=0.2, random_state=42)

eta = RandomForestRegressor(n_estimators=200, random_state=42, n_jobs=-1)
eta.fit(Xtr, ytr)
pred = eta.predict(Xte)
mae = mean_absolute_error(yte, pred)
rmse = float(np.sqrt(mean_squared_error(yte, pred)))
r2 = r2_score(yte, pred)
print(f"ETA RF200 — MAE {mae:.2f} min  |  RMSE {rmse:.2f} min  |  R2 {r2:.4f}")
print("Feature importance:", dict(zip(FEATURES, np.round(eta.feature_importances_, 4))))

plt.figure(figsize=(5,5))
plt.scatter(yte, pred, s=6, alpha=0.35)
lim = [min(yte.min(), pred.min()), max(yte.max(), pred.max())]
plt.plot(lim, lim, "r--")
plt.xlabel("actual (min)"); plt.ylabel("predicted (min)"); plt.title("ETA: aktual vs prediksi"); plt.tight_layout(); plt.show()

# contoh 1 rute: hitung ETA dari ETD
ex = dte.iloc[0]
etd = pd.to_datetime(ex["etd"])
feat = np.array([[ex["distance_km"], ex["cruise_speed_kmh"], ex["avg_wind_kmh"], ex["weather_severity"]]])
pmin = float(eta.predict(feat)[0])
print(f"Contoh: {ex['tail_number']} {ex['origin_code']}->{ex['dest_code']}  {ex['distance_km']} km  "
      f"ETD {etd}  pred {pmin:.1f} min  -> ETA {etd+pd.to_timedelta(pmin,'m')}  "
      f"(aktual {ex['actual_duration_minutes']} min, cuaca {ex['origin_weather']}/{ex['dest_weather']})")
````

## 2. Delay classifier — apakah delay > 15 menit?

````python
from sklearn.linear_model import LogisticRegression

df["is_delayed"] = (df["delay_minutes"] > 15).astype(int)
print(df["is_delayed"].value_counts().to_dict(), "— delayed rate", round(df["is_delayed"].mean()*100,1), "%")
Xd = df[FEATURES].values
yd = df["is_delayed"].values
Xtr2, Xte2, ytr2, yte2 = train_test_split(Xd, yd, test_size=0.2, random_state=42, stratify=yd)
clf = RandomForestClassifier(n_estimators=200, random_state=42, n_jobs=-1)
clf.fit(Xtr2, ytr2)
yp = clf.predict(Xte2)
print(classification_report(yte2, yp, digits=3))
print("Confusion matrix:\n", confusion_matrix(yte2, yp))
````

## 3. Bottleneck — `bottleneck_train.csv` (fitur sama: predictive_model.py)

````python
bn = pd.read_csv(DATA / "bottleneck_train.csv")
bn.head()
````

````python
Fb = ["avg_flight_hours","refueler_trucks","forklifts","deployed_aircraft"]
Xb, yb = bn[Fb].values, bn["is_critical"].values
Xtr3, Xte3, ytr3, yte3 = train_test_split(Xb, yb, test_size=0.2, random_state=42, stratify=yb)
rfb = RandomForestClassifier(n_estimators=150, random_state=42)
rfb.fit(Xtr3, ytr3)
print(classification_report(yte3, rfb.predict(Xte3), digits=3))
print(dict(zip(Fb, np.round(rfb.feature_importances_,4))))
````
